```python
import yfinance as yf
import pandas as pd
import  numpy as np
from datetime import datetime, timedelta
from scipy.signal import find_peaks, peak_prominences
```

```python
def get_stock_data(ticker, start_date=None, end_date=None):

    if start_date is None:
        start_date = (datetime.now() - timedelta(days=365)).strftime('%Y-%m-%d')
    if end_date is None:
        end_date = datetime.now().strftime('%Y-%m-%d')


        # Ensure ticker is a string
    stock_data = yf.download(ticker, start=start_date, end=end_date, progress=True, auto_adjust=True)
    if stock_data.empty:
        raise Exception(f"No data found for ticker: {ticker}")

    # Ensure the index is a datetime index

    # IMPORTANT FIX: Sort data in ascending order for correct time-series calculations
    stock_data.sort_values(by='Date', ascending=True, inplace=True)

    stock_data = stock_data.droplevel(level=1, axis=1)
    stock_data['Ticker'] = ticker


   # stock_data['20_DMA'] = calculate_dma(stock_data['Close'], window=20)
    return stock_data

def get_stock_list():
    #Returns a list of popular stock ticker symbols.

    Returns: 
    list: A list of stock ticker symbols
    #getting 10 Nifty 50 Stock data
    stocks = ["APOLLOHOSP", "NESTLEIND", "ULTRACEMCO", "HDFCLIFE", "BAJFINANCE", "ITC", "SHRIRAMFIN", "TATASTEEL", "TCS", "MARUTI"]

    return stocks
```
